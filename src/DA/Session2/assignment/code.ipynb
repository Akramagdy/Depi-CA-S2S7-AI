```python
import pandas as pd
df = pd.read_csv('insurance.csv')
```

```python
df.head
```

```python
df = df.drop
```

```python
dtype = df.dtypes
nunique = df.nunique()
pd.DataFrame({'dtype': dtype, 'nunique': nunique}).T
```

```python
cols = ["sex", "children", "smoker", "region"]
df[cols] = df[cols].astype("category")

pd.DataFrame({"dtype": df.dtypes, "nunique": df.nunique()}).T
```

```python
null = df.isnull().sum()
ratio = null / len(df)
pd.DataFrame({"null": null, "ratio": ratio}).T
```
