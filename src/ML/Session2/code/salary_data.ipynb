```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, root_mean_squared_error
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('Salary_Data.csv')
```

```python
plt.figure(figsize= (3,3))
sns.pairplot(data, x_vars= ['YearsExperience'], y_vars=['Salary'], size= 3, kind= "scatter")
plt.title("salary vs years")
plt.xlabel("years")
plt.ylabel("salary")
plt.show()
```

```python
x = data.iloc[:,:-1]
y = data.iloc[:,1]
x_train, x_test, y_train, y_test = train_test_split(x,y, train_size= 0.8, random_state= 20)
x_train
```

```python
reg_line = LinearRegression()
reg_line.fit(x_train, y_train)
#w = reg_line.coef_
#b = reg_line.intercept_
y_hat_train = reg_line.predict(x_train)
y_hat_test = reg_line.predict(x_test)
```

```python
plt.scatter(x_train, y_train, color = 'red')
plt.plot(x_train, y_hat_train, color ='blue')
plt.title("Salary vs Experience (Training set)")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.show()
```

```python
#calculate mean sqaured error
mse_train = mean_squared_error(y_train, y_hat_train)
mse_test = mean_squared_error(y_test,y_hat_test)
print(mse_train)
print(mse_test)
rmse = root_mean_squared_error(y_test, y_hat_test)
r2 = r2_score(y_test, y_hat_test)
r2
```
