```python
import numpy as np
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score, mean_squared_error
import matplotlib.pyplot as plt
```

```python
x = np.array([[1], [2], [3], [4]])
y = np.array([2, 4, 6, 8])

#linear regression
lin_reg = LinearRegression()
lin_reg.fit(x,y)
y_pred_lin = lin_reg.predict(x)

#ridge regression
ridge = Ridge(alpha=1)
ridge.fit(x,y)
y_pred_ridge = ridge.predict(x)

#Lasso Regression
lasso = Lasso(alpha=1.0)
lasso.fit(x,y)
y_pred_lasso = lasso.predict(x)
```

```python
print("LinearRegression")
print("mse: ", mean_squared_error(y , y_pred_lin))
print("w : ", lin_reg.coef_)
print("b : ", lin_reg.intercept_)
print("="*30)
 
print("Ridge Regression")
print("mse: ", mean_squared_error(y , y_pred_ridge))
print("w : ", ridge.coef_)
print("b : ", ridge.intercept_)
print("="*30)
 
print("lasso Regression")
print("mse: ", mean_squared_error(y , y_pred_lasso))
print("w : ", lasso.coef_)
print("b : ", lasso.intercept_)
print("="*30)
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.scatter(x,y)
plt.plot(x,y_pred_lin, color="red")
plt.title("Linear Regression")
```

```python
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.scatter(x,y,  label ="Actual Data")
plt.plot(x,y_pred_lin, color="red", label ="Linear Regression")
plt.plot(x,y_pred_ridge, color="green",linestyle="--", label ="Linear Regression")
plt.plot(x,y_pred_lasso, color="purple",linestyle=":", label ="Linear Regression")
 
plt.title("Linear Regression")
plt.legend()
plt.show()
```
