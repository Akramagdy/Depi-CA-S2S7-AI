```python
# Task 1
email = "Amit_ml@gmail.edu"

if email.count("@") != 1 or "." not in email.split("@")[1]:
    print("Invalid email")
else:
    username = email.split("@")[0]
    domain_part = email.split("@")[1]
    domain = domain_part.rsplit(".", 1)[0]

    if email.endswith(".com"):
        domain_type = "Commercial Domain"
    elif email.endswith(".edu"):
        domain_type = "Educational Domain"
    else:
        domain_type = "Other Domain"

    print(f"Username: {username}")
    print(f"Domain: {domain}")
    print(f"Domain Type: {domain_type}")
```

```python
# Task 2
enc_message = "###!!@mocleW EPGTQ!!!6789"
cleaned = "".join(filter(str.isalpha, enc_message))

p1 ,p2 = cleaned.split("E")

p1 = p1[::-1] 

new = f"{p1} {p2}"
print(new)

import re

encoded = "###!!@mocleW EPGTQ!!!6789"

core = re.search(r"[a-zA-Z]+ [a-zA-Z]+", encoded).group()
first_word, second_word = core.split()

first_word = first_word[::-1]  # 'Welcom'

decoded_message = f"{first_word} {second_word}"
print(decoded_message)
```

```python
# Task 3
import re

encoded = "&&&**$gnirtS PLIO!!@1234"

f1, s2 = re.findall(r"[a-zA-Z]+", encoded)
f1 = f1[::-1]
vowel_map = str.maketrans("IO", "EU")
s2 = s2.translate(vowel_map)

decoded_message = f"{f1} {s2}"
print(decoded_message)
```

```python
# Task 4
import re

enc_message = "##$$$@!yalpstcejorp EPUVT****9887"

word1, word2 = re.findall(r"[a-zA-Z]+", enc_message)

word1 = word1[::-1]  
vowel_map = str.maketrans({"E": "A", "U": "O"})
word2 = word2.translate(vowel_map)

decoded_message = f"{word1} {word2}"
print(decoded_message)
```
