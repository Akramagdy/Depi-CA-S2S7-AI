```python
def is_prime(number :int|None = 1):
    '''
    Determines whether a given integer is a prime number and prints the result.
    Args:
            number(int): The integer to evaluate for primality. 
            Defaults to 1.
    
    Returns:
        None: This function prints the output directly and does not return a value.
    '''

    is_prime = True
 
    if number <= 1:
        is_prime = False
    else:
        for i in range(2, int(number ** 0.5) + 1):
            if number % i == 0:
                is_prime = False
                break
    
    if is_prime:
        print(f"The number {number} is prime.")
    else:
        print(f"The number {number} is not prime.")
```

```python
def train_model(model_name, *datasets, **hyperparameters):
    print(f"Training Model: {model_name}")

    print("\nDatasets")
    for dataset in datasets:
        print("-", dataset)

    print("\nHyperparameters")
    for key, value in hyperparameters.items():
        print(f"{key}: {value}")
```

```python
train_model(
    "Fraud Detection",
    "train.csv",
    "validation.csv",
    "test.csv",
    learning_rate =0.001,
    epochs =100,
    batch_size =64,
    optimizer = "Adam",
    early_stopping =True
)
```
