```python
def is_prime(number :int|None = 1) -> bool:
    '''
    Determines whether a given integer is a prime number and prints the result.
    Args:
            number(int): The integer to evaluate for primality. 
            Defaults to 1.
    
    Returns:
        1: if number is prime
        0: if number is not prime
    '''

    is_prime = True
 
    if number <= 1:
        is_prime = False
    else:
        for i in range(2, int(number ** 0.5) + 1):
            if number % i == 0:
                is_prime = False
                break
    
    if is_prime:
        return 1
    else:
        return 0


def twin_primes(upper : int |None = 5) -> None:
    '''
        Determines twin primes in a specified range and prints the result.
        Args:
                upper(int): The integer to specify range. 
                Defaults to 5.
        
        Returns:
            None: This function prints the output directly and does not return a value.
        '''
    
    print("Twin Prime:")
    for i in range(3,upper -1,2):
        if is_prime(i) and is_prime(i+2):
            print(f"{i} and {i+2}")
```
