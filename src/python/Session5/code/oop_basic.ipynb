```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def move(self):
        pass

class Bird(Animal):
    def move(self):
        print("move from Bird")


class Cat(Animal):
    def move(self):
         print("move from Cat")

a = Bird()
a.move()
b = Cat()
b.move()
```

```python
class Shape:
    def calculate_area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return 3.14 * (self.radius)**2

class Square(Shape):
    def __init__(self, side):
        self.side =side

    def calculate_area(self):
        return self.side**2

```

```python
class Model:

    def fit(self, x,y):
        raise NotImplementedError
    def predict(self, x):
        raise NotImplementedError

class LinearRegression(Model):
    def fit(self, x, y):
        print("fitting linear regression")

    def predict(self,x):
        print("predicting with linear regression")

class DicisionTree(Model):
    def fit(self, x, y):
        print("fitting Decision Tree")

    def predict(self, x):
        print("predicting with Decision Tree")

models = [LinearRegression(), DicisionTree()]
for model in models:
    model.fit(None, None)
    model.predict(None)
```

```python
def array_reshape(arr : list, rows:int =-1, cols: int =-1):
    try:
        arr = np.array(arr)
        return arr.reshape(rows,cols)
    except(TypeError, ValueError) as e:
        raise ValueError(f"Invalid reshape operation: {e}")
```

```python
vec = [1,2,3,4,5,6]
array_reshape(vec,2,3)
```
