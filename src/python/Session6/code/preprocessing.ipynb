```python
def gen3x3mat(shape, n=8):
    return (n *np.random.random(shape)).astype(int)
```

```python
x = gen3x3mat((4,3), 43)
print(x)
```

```python
mat1 = np.ones((3,4))
print(mat1)
print("="*20)
mat2 = np.zeros((5,3))
print(mat2)
np.savez('np_file.npz', cat = mat1, dog = mat2)
 
```
